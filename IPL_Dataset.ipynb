```python
team_season_matches = (
    pd.concat([
        matches[['season','team1']].rename(columns={'team1':'team'}),
        matches[['season','team2']].rename(columns={'team2':'team'})
    ]).groupby(['season','team']).size().reset_index(name='matches')
)
```

```python
team_season_matches.head(10)
```

```python
team_season_wins = (
    matches.groupby(['season','winner'])
    .size()
    .reset_index(name='wins')
)
```

```python
team_season_wins.head(10)
```

```python
team_season_stats = team_season_matches.merge(
    team_season_wins,
    left_on =['season','team'],
    right_on = ['season','winner'],
    how = 'left'
)
```

```python
team_season_stats.head(10)
```

```python
team_season_stats = team_season_stats.drop(columns='winner')
```

```python
team_season_stats.head(10)
```

```python
team_season_stats['win_rate'] = (
    team_season_stats['wins']/team_season_stats['matches']*100
)
```

```python
team_season_stats['win_rate'] = team_season_stats['win_rate'].round(2)
team_season_stats.head(10)
```

```python
selected_teams = [
    'Chennai Super Kings',
    'Mumbai Indians',
    'Royal Challengers Bengaluru'
]

season_performance = team_season_stats[
    team_season_stats['team'].isin(selected_teams)
]
```

```python
season_performance
```

```python
teams = team_season_stats['team'].unique()
```

```python
teams
```

```python
buttons = []
for team in teams:
  buttons.append({
      'label':team,
      'method':'update',
      'args':[
          {'visible':[t == team for t in teams]},
          {'title':f'{team}-Season-wise Win Rates'}
      ]
  })
```

```python
import plotly.graph_objects as go

fig = go.Figure()

for team in teams:
  team_data = team_season_stats[
      team_season_stats['team'] == team
  ]

  fig.add_trace(
      go.Scatter(
          x=team_data['season'],
          y=team_data['win_rate'],
          mode='lines+markers',
          name = team,
          visible=(team == teams[0])
      )
  )
```

```python
fig.update_layout(
    updatemenus=[
        dict(
            buttons=buttons,
            direction = 'down',
            showactive = True,
            x=1.02,
            y=1
        )
    ],
    title = f'{teams[0]} - Season_wise Win Rate',
    xaxis_title = 'Season',
    yaxis_title = 'Win Rate (%)',
    yaxis=dict(range=[0,100])
)

fig.show()
```
